2. Importación de librerías y verificación del entorno

```python
import sys
from pathlib import Path

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, r2_score

print("Python:", sys.version.split()[0])
print("Pandas:", pd.__version__)
print("NumPy:", np.__version__)
print("Entorno listo para el análisis.")
```

3. Lectura del dataset
El archivo debe estar en la misma carpeta del notebook.

```python
RUTA = Path("datos_produccion_lechera_kaggle_mejorada.csv")

if not RUTA.exists():
    raise FileNotFoundError(
        "No se encontró el archivo. Coloque 'datos_produccion_lechera_kaggle_mejorada.csv' "
        "en la misma carpeta del notebook."
    )

df = pd.read_csv(RUTA)

print(f"Filas: {df.shape[0]}")
print(f"Columnas: {df.shape[1]}")
display(df.head())
```

4. Reconocimiento de estructura y variables
El dataset contiene variables descriptivas del animal y del entorno, una variable objetivo de producción de leche y una secuencia de 100 mediciones de ordeño.

```python
print("Columnas principales:")
columnas_principales = [
    "nombre_vaca", "peso_animal", "edad", "dias_lactancia",
    "temperatura", "humedad_ubre", "alimentacion",
    "raza_Holstein", "produccion_leche"
]
display(df[columnas_principales].head())

print("\nTipos de datos:")
display(df[columnas_principales].dtypes.to_frame("tipo"))

columnas_ordeno = [c for c in df.columns if c.startswith("ordeño_")]
print(f"Variables secuenciales de ordeño detectadas: {len(columnas_ordeno)}")
```

5. Calidad de datos: nulos y duplicados
Primero se verifica el estado real del dataset. Si no existen valores nulos, se conserva ese resultado y se demuestra el procedimiento de limpieza en una copia controlada, sin alterar los datos originales.

```python
nulos_por_columna = df.isna().sum()
total_nulos = int(nulos_por_columna.sum())
duplicados = int(df.duplicated().sum())

print("Total de valores nulos:", total_nulos)
print("Filas duplicadas:", duplicados)

if total_nulos > 0:
    print("\nColumnas con nulos:")
    display(nulos_por_columna[nulos_por_columna > 0].sort_values(ascending=False))
```

```python
# Procedimiento robusto de limpieza sobre el dataset real
filas_antes = len(df)
df_limpio = df.dropna().drop_duplicates().copy()
filas_despues = len(df_limpio)

print(f"Filas antes de limpiar: {filas_antes}")
print(f"Filas después de limpiar: {filas_despues}")
print(f"Filas eliminadas: {filas_antes - filas_despues}")
print("Nulos restantes:", int(df_limpio.isna().sum().sum()))
```

Demostración explícita del tratamiento de nulos
Como este dataset puede venir completamente limpio, se crea una copia de demostración con un valor nulo artificial para evidenciar el procedimiento dropna() sin modificar el dataset original.

```python
demo_nulos = df_limpio[["nombre_vaca", "humedad_ubre", "produccion_leche"]].head(5).copy()

# Solo para demostración académica:
demo_nulos.loc[demo_nulos.index[0], "humedad_ubre"] = np.nan

print("Antes:")
display(demo_nulos)

demo_sin_nulos = demo_nulos.dropna()

print("Después de eliminar registros con nulos:")
display(demo_sin_nulos)
```

6. Cálculos estadísticos básicos
Se selecciona produccion_leche como una variable numérica central para el análisis.

```python
serie = df_limpio["produccion_leche"]

resumen = pd.Series({
    "media": serie.mean(),
    "mediana": serie.median(),
    "desviacion_estandar": serie.std(),
    "minimo": serie.min(),
    "maximo": serie.max()
}).round(3)

display(resumen.to_frame("valor"))
```

7. Ordenación de datos
Se identifican los animales con mayor producción registrada.

```python
ranking = (
    df_limpio[
        ["nombre_vaca", "peso_animal", "edad", "produccion_leche"]
    ]
    .sort_values("produccion_leche", ascending=False)
    .head(10)
)

display(ranking)
```

8. Agrupamiento
La variable raza_Holstein está codificada como 1/0. Se crea una etiqueta descriptiva y se comparan promedios.

```python
df_limpio["tipo_raza"] = np.where(
    df_limpio["raza_Holstein"].eq(1),
    "Holstein",
    "No Holstein"
)

resumen_raza = (
    df_limpio
    .groupby("tipo_raza")
    .agg(
        cantidad=("nombre_vaca", "count"),
        produccion_promedio=("produccion_leche", "mean"),
        peso_promedio=("peso_animal", "mean"),
        humedad_promedio=("humedad_ubre", "mean")
    )
    .round(2)
    .sort_values("produccion_promedio", ascending=False)
)

display(resumen_raza)
```

9. Visualización 1 - Distribución de la producción
El histograma permite observar concentración, dispersión y forma general de una variable numérica.

```python
plt.figure(figsize=(9, 5))
plt.hist(df_limpio["produccion_leche"], bins=15, edgecolor="black")
plt.title("Distribución de la producción de leche")
plt.xlabel("Producción de leche")
plt.ylabel("Frecuencia")
plt.grid(axis="y", alpha=0.25)
plt.show()
```

10. Visualización 2 - Relación entre peso y producción
El gráfico de dispersión es adecuado para explorar la relación entre dos variables numéricas.

```python
plt.figure(figsize=(9, 5))
plt.scatter(
    df_limpio["peso_animal"],
    df_limpio["produccion_leche"],
    alpha=0.75
)
plt.title("Relación entre peso animal y producción de leche")
plt.xlabel("Peso del animal")
plt.ylabel("Producción de leche")
plt.grid(alpha=0.25)
plt.show()

correlacion = df_limpio["peso_animal"].corr(df_limpio["produccion_leche"])
print(f"Correlación peso-producción: {correlacion:.4f}")
```

11. Visualización 3 - Serie de ordeños de un animal
El dataset contiene 100 mediciones de ordeño por animal. Se visualiza la secuencia para un caso individual.

```python
vaca = df_limpio.iloc[0]
valores_ordeno = vaca[columnas_ordeno].astype(float).to_numpy()

plt.figure(figsize=(11, 5))
plt.plot(range(1, len(valores_ordeno) + 1), valores_ordeno, marker="o", markersize=3)
plt.title(f"Evolución de 100 ordeños - {vaca['nombre_vaca']}")
plt.xlabel("Número de ordeño")
plt.ylabel("Producción registrada")
plt.grid(alpha=0.25)
plt.show()
```

12. Bonus técnico - Modelo predictivo de IA
Este bloque es opcional para el video. Muestra cómo el análisis exploratorio puede dar paso a un modelo de Machine Learning.
Criterio técnico importante: se excluyen las 100 columnas ordeño_* para evitar introducir variables que podrían representar información demasiado cercana al resultado final y generar fuga de información (data leakage).
El objetivo no es presentar este modelo como un sistema listo para producción, sino demostrar el flujo de entrenamiento y evaluación.

```python
features = [
    "peso_animal",
    "edad",
    "dias_lactancia",
    "temperatura",
    "humedad_ubre",
    "alimentacion",
    "raza_Holstein"
]

X = df_limpio[features]
y = df_limpio["produccion_leche"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

modelo = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)

modelo.fit(X_train, y_train)
predicciones = modelo.predict(X_test)

mae = mean_absolute_error(y_test, predicciones)
r2 = r2_score(y_test, predicciones)

print(f"MAE: {mae:.4f}")
print(f"R²: {r2:.4f}")
```

```python
importancias = (
    pd.Series(modelo.feature_importances_, index=features)
    .sort_values(ascending=True)
)

plt.figure(figsize=(9, 5))
plt.barh(importancias.index, importancias.values)
plt.title("Importancia de variables - Random Forest")
plt.xlabel("Importancia relativa")
plt.ylabel("Variable")
plt.show()

display(importancias.sort_values(ascending=False).to_frame("importancia"))
```
